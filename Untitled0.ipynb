```python
# =====================================================================
# Polynomial Regression Assignment - BT2024131
#
# var1 -> Lasso, degree <= 10
# var2 -> Ridge, degree <= 20
#
# FAST + NO DATA LEAKAGE VERSION
# =====================================================================

import numpy as np
import pandas as pd
import warnings
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.linear_model import LassoCV, RidgeCV
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold, cross_val_predict
from sklearn.metrics import mean_squared_error, r2_score


# =====================================================================
# SETTINGS
# =====================================================================

ROLL = "BT2024131"

DEGREES = {
    1: range(1, 11),
    2: range(1, 21)
}

METHODS = {
    1: "Lasso",
    2: "Ridge"
}

N_FOLDS = 5
RANDOM_STATE = 0

# Smaller grids = much faster
LASSO_ALPHAS = np.logspace(-3, 0, 10)
RIDGE_ALPHAS = np.logspace(-2, 4, 12)


# =====================================================================
# CREATE PIPELINE
# =====================================================================

def make_model(method, degree):

    if method == "Lasso":

        reg = LassoCV(
            alphas=LASSO_ALPHAS,
            cv=3,
            max_iter=5000,
            tol=1e-3,
            n_jobs=-1,
            random_state=RANDOM_STATE
        )

    else:

        reg = RidgeCV(
            alphas=RIDGE_ALPHAS,
            cv=3
        )

    return Pipeline([
        (
            "poly",
            PolynomialFeatures(
                degree=degree,
                include_bias=False
            )
        ),
        (
            "scale",
            StandardScaler()
        ),
        (
            "reg",
            reg
        )
    ])


# =====================================================================
# RUN ONE VARIABLE
# =====================================================================

def run(var):

    method = METHODS[var]
    degrees = list(DEGREES[var])

    train_file = f"{ROLL}_train_var{var}.csv"
    test_file = f"{ROLL}_test_var{var}.csv"

    train_df = pd.read_csv(train_file)
    test_df = pd.read_csv(test_file)

    features = [
        c for c in train_df.columns
        if c != "y"
    ]

    X = train_df[features].values
    y = train_df["y"].values

    X_test = test_df[features].values

    has_test_y = "y" in test_df.columns

    if has_test_y:
        y_test = test_df["y"].values


    # ================================================================
    # HEADING
    # ================================================================

    print("\n")
    print("=" * 90)
    print(f"                              VAR{var}")
    print("=" * 90)

    print(f"Method           : {method}")
    print(f"Features         : {len(features)}")
    print(f"Train samples    : {len(X)}")
    print(f"Test samples     : {len(X_test)}")
    print(f"Degree range     : {degrees[0]} - {degrees[-1]}")
    print(f"CV folds         : {N_FOLDS}")


    # ================================================================
    # CROSS VALIDATION
    # ================================================================

    kf = KFold(
        n_splits=N_FOLDS,
        shuffle=True,
        random_state=RANDOM_STATE
    )

    results = []

    best_degree = None
    best_mse = float("inf")
    best_oof = None


    print("\n")
    print("-" * 90)
    print("DEGREE SELECTION")
    print("-" * 90)

    print(
        f"{'Degree':>8}"
        f"{'Train MSE':>15}"
        f"{'Train R2':>13}"
        f"{'Validation MSE':>20}"
        f"{'Validation R2':>18}"
        f"{'Alpha':>14}"
    )

    print("-" * 90)


    # ================================================================
    # TEST EACH DEGREE
    # ================================================================

    for degree in degrees:

        model = make_model(
            method,
            degree
        )

        # ------------------------------------------------------------
        # 5-fold out-of-fold predictions
        # ------------------------------------------------------------

        oof = cross_val_predict(
            model,
            X,
            y,
            cv=kf,
            n_jobs=-1
        )

        validation_mse = mean_squared_error(
            y,
            oof
        )

        validation_r2 = r2_score(
            y,
            oof
        )


        # ------------------------------------------------------------
        # Fit on all training data
        # only for train metrics
        # ------------------------------------------------------------

        model.fit(X, y)

        train_pred = model.predict(X)

        train_mse = mean_squared_error(
            y,
            train_pred
        )

        train_r2 = r2_score(
            y,
            train_pred
        )

        alpha = model.named_steps["reg"].alpha_


        # ------------------------------------------------------------
        # Save results
        # ------------------------------------------------------------

        results.append({
            "Degree": degree,
            "Train MSE": train_mse,
            "Train R2": train_r2,
            "Validation MSE": validation_mse,
            "Validation R2": validation_r2,
            "Alpha": alpha
        })


        # ------------------------------------------------------------
        # Best model based ONLY on validation MSE
        # ------------------------------------------------------------

        if validation_mse < best_mse:

            best_mse = validation_mse
            best_degree = degree
            best_oof = oof


        print(
            f"{degree:>8}"
            f"{train_mse:>15.5f}"
            f"{train_r2:>13.5f}"
            f"{validation_mse:>20.5f}"
            f"{validation_r2:>18.5f}"
            f"{alpha:>14.6g}"
        )


    results_df = pd.DataFrame(results)

    best_row = results_df[
        results_df["Degree"] == best_degree
    ].iloc[0]


    # ================================================================
    # BEST MODEL
    # ================================================================

    print("\n")
    print("=" * 90)
    print(f"                         BEST MODEL - VAR{var}")
    print("=" * 90)

    print(f"Model              : {method}")
    print(f"Selected degree    : {best_degree}")
    print(f"Selected alpha     : {best_row['Alpha']:.6g}")


    # ================================================================
    # VALIDATION
    # ================================================================

    print("\n")
    print("-" * 90)
    print("VALIDATION PERFORMANCE")
    print("-" * 90)

    print(
        f"Validation MSE     : "
        f"{best_row['Validation MSE']:.6f}"
    )

    print(
        f"Validation R2      : "
        f"{best_row['Validation R2']:.6f}"
    )


    # ================================================================
    # TRAINING
    # ================================================================

    print("\n")
    print("-" * 90)
    print("TRAINING PERFORMANCE")
    print("-" * 90)

    print(
        f"Training MSE       : "
        f"{best_row['Train MSE']:.6f}"
    )

    print(
        f"Training R2        : "
        f"{best_row['Train R2']:.6f}"
    )


    # ================================================================
    # OVERFITTING CHECK
    # ================================================================

    ratio = (
        best_row["Validation MSE"]
        / best_row["Train MSE"]
    )

    print("\n")
    print("-" * 90)
    print("OVERFITTING CHECK")
    print("-" * 90)

    print(
        f"Validation / Train MSE : {ratio:.3f}"
    )

    if ratio < 1.25:
        print("Diagnosis              : Very little overfitting")

    elif ratio < 2.0:
        print("Diagnosis              : Some overfitting, acceptable")

    else:
        print("Diagnosis              : Significant overfitting")


    # ================================================================
    # FINAL MODEL
    # ================================================================

    final_model = make_model(
        method,
        best_degree
    )

    final_model.fit(X, y)

    test_predictions = final_model.predict(
        X_test
    )


    # ================================================================
    # SAVE PREDICTIONS
    # ================================================================

    output_file = (
        f"{ROLL}_pred_var{var}.csv"
    )

    pd.DataFrame({
        "y": test_predictions
    }).to_csv(
        output_file,
        index=False
    )


    # ================================================================
    # TEST PERFORMANCE
    # ================================================================

    print("\n")
    print("-" * 90)
    print("TEST PERFORMANCE")
    print("-" * 90)

    if has_test_y:

        test_mse = mean_squared_error(
            y_test,
            test_predictions
        )

        test_r2 = r2_score(
            y_test,
            test_predictions
        )

        print(
            f"Test MSE           : "
            f"{test_mse:.6f}"
        )

        print(
            f"Test R2            : "
            f"{test_r2:.6f}"
        )

    else:

        print("Test labels not available.")
        print("Test MSE/R2 cannot be calculated.")
        print(
            f"Predictions generated : "
            f"{len(test_predictions)}"
        )


    print(
        f"\nPrediction file    : {output_file}"
    )


    # ================================================================
    # PLOT
    # ================================================================

    plt.figure(figsize=(8, 5))

    plt.plot(
        results_df["Degree"],
        results_df["Train MSE"],
        "o--",
        label="Training MSE"
    )

    plt.plot(
        results_df["Degree"],
        results_df["Validation MSE"],
        "o-",
        label="Validation MSE"
    )

    plt.axvline(
        best_degree,
        linestyle=":",
        label=f"Selected degree = {best_degree}"
    )

    plt.xlabel("Polynomial Degree")
    plt.ylabel("MSE")

    plt.title(
        f"var{var}: {method} Degree Selection"
    )

    plt.yscale("log")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()

    plot_file = f"results_var{var}.png"

    plt.savefig(
        plot_file,
        dpi=150
    )

    plt.show()

    print(f"Plot saved         : {plot_file}")

    print("=" * 90)

    return results_df


# =====================================================================
# RUN VAR1 FIRST
# =====================================================================

results_var1 = run(1)


# =====================================================================
# RUN VAR2 SECOND
# =====================================================================

results_var2 = run(2)


# =====================================================================
# DOWNLOAD — UNCOMMENT WHEN NEEDED
# =====================================================================
from google.colab import files

files.download(f"{ROLL}_pred_var1.csv")
files.download(f"{ROLL}_pred_var2.csv")
files.download("results_var1.png")
files.download("results_var2.png")
```
